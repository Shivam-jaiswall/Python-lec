```python
# The if Statement: Making Decisions ########
# We use if when we want our code to make a choice. It allows a specific block of code to run only once, and only if a certain condition is True.
# Think of it like this: "If it is raining outside, take an umbrella."

# Example

weather = "raining"

if weather == "raining":
    print("Take an umbrella!") # This runs because the condition is true
```

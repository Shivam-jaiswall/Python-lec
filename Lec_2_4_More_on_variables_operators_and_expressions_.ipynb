                                                              **IN OPERATORS**

```python
print('alpha' in 'A variable name can only contain alpha-numeric characters and underscores')
# It checks if a specific sequence of characters exists seamlessly within a larger text.
# It is strictly case-sensitive. For example, 'apple' in 'pineapple' is True, but 'Apple' in 'pineapple' is False

print('alpha' in 'A variable must start with a letter r the underscore characters')
# in this case 'alpha' is not in the senetence
```

```python
x=5
print(1 < x < 10)
print(10 < x < 20)
print(x < 10 < x * 10 < 100)
print(10 > x <= 9)
print(5 == x > 4)
```

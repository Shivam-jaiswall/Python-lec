```python
# Let us find the factorial of a number

print("enter a number")
n = int(input())

# n= 5
# 1*2*3*4*5 = 120

i = 1
answer = 1
while (i <= n ): # Less than or equal to
  answer = answer * i
  i = i + 1

print(answer)
```

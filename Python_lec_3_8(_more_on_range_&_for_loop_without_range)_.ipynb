```python
for x in range(1, 11, 2):  # (First parameter is start, Second is stop, Third is step)
  print(x)
```

```python
for x in range(1, 25, 3):
  print(x)
```

```python
for x in range(9, -1, -1): # In reverse order
  print(x)
```

```python
for x in range(10):
  print(x, end = " ") # (end = " ") - "When you finish printing the number, do not jump to a new line. Instead, just put a blank space at the end."
```

```python
d = 3
m = 8
y = 2026
print("Today's date is", end = " ")
print(d, m, y, sep = "/") # (sep = "/ ") - The sep instruction only does something when you are trying to print two or more items at the exact same time
```

```python
n = int(input())
for i in range(1,11):
 # print(n, "X", i, "=", n*i )

 # print(f"{n} X {i} = {n*i}")

 # print("{0} X {1} = {2}".format(n, i, n*i))  (Here 0 is replaced by n), (1 replaced by i) and (2 is replaced by n*i)

   print("%d X %d = %d" % (n, i , n*i) ) # here d stands for integers
```

```python
pi = 22/7
print(f"Value Of Pie = {pi:.3f}")
print("Value Of Pie = {0:.6f}".format(pi))
print("Value of Pie = %.5f" %(pi))
```

```python
print("{0}".format(1))
print("{0}".format(11))
print("{0}".format(111))
print("{0}".format(1111))
print("{0}".format(11111))
```

```python
print("{0:5d}".format(1))
print("{0:5d}".format(11))
print("{0:5d}".format(111))
print("{0:5d}".format(1111))
print("{0:5d}".format(11111))
```
